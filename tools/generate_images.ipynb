```
import torch, os
from diffusers import AutoPipelineForText2Image
pipe = AutoPipelineForText2Image.from_pretrained("stabilityai/sdxl-turbo", torch_dtype=torch.float16, variant="fp16").to("cuda")
```

```
os.makedirs("destinations", exist_ok=True)
for dest, country, landmark, slug in ITEMS:
    prompt = f"{landmark} in {dest}, {country}, scenic travel photograph, golden hour, wide angle, vivid colors, no text, no people"
    img = pipe(prompt, num_inference_steps=2, guidance_scale=0.0, width=512, height=512).images[0]
    img.convert("RGB").save(f"destinations/{slug}.jpg", quality=88)
    print("done", slug)
```
